```python
class Employee:
    a=1
    @classmethod
    def show(self):
        print(f"the value of a is {self.a}")
        
        
e=Employee()
e.a=45
print(e.a)

e.show()
        
 





```

```python
class Employee:
    a=1
    @classmethod
    def show(self):
        print(f"the value of a is {self.a}")
        
    @property
    def name(self):
        return self.name
e=Employee()
e.a=45
print(e.a)

e.show()
```

```python
class TwoDVector:
    def __init__(self,i,j):
        self.i=i
        self.j=j
    def show(self):
        print(f'the value is: {self.i}i +{self.j}j')


class  ThreeDVector(TwoDVector):
    def __init__(self,i,j,k):
        super().__init__(i,j)
        
        
     
        self.k=k  
    def show(self):
        print(f'the vector is :{self.i}i +{self.j}+{self.k}k') 
        
a=TwoDVector(1,2)
b=ThreeDVector(1,2,3)
a.show()
b.show()        
        
  
```

```python
def http_status(status):
    match status:
        case 200:
            return "okk"
        case 404:
            return "Not Found"
        case 500:
            return "Internal Server Error"
        
        case _:
            return "Unknown Status"
        
print(http_status(5007))
```
