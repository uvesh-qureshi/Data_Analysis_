```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
import numpy as np
x=np.array([[2],[4],[6],[8],[10],[12],[14]])
y=np.array(['20000','40000','55000','85000','100000','120000','140000'])
lr.fit(x,y)
loef=lr.coef_[0]
loef
```

```python
intrcept=lr.intercept_
intrcept
```

```python

import numpy 

x=(2,4,6,8,10,12,14)
y=numpy.array([20000,50000,55000,60000,85000,100000,120000])
```

```python
mymodel=numpy.poly1d(numpy.polyfit(x,y,1))
pred=mymodel(x)
pred
```

```python
import matplotlib.pyplot as plt

plt.scatter(x,y)

plt.plot(x,pred)
plt.show()
```

```python
from sklearn.metrics import r2_score
print(r2_score(y,pred))
```
