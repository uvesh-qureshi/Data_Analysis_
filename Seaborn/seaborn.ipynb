```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
flight=sns.load_dataset('flights')
```

```python
tips=sns.load_dataset('tips')
tips
```

```python
tipscorr=tips[['total_bill','tip','size']]
tipscorr
```

```python
tipscorr.corr()
```

```python
sns.heatmap(tipscorr.corr(),annot=True)
```
